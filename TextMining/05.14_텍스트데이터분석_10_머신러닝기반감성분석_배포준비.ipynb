```python
# 한국어 토크나이저 정의
from konlpy.tag import Okt
def korean_tokenizer(text):
    my_tags = ['Noun', 'Verb', 'Adjective'] # 명사, 동사, 형용사만 추출
    my_stopwords = [] # 불용어 제거
    tokenizer = Okt().pos
    return [word for word, tag in tokenizer(text) if tag in my_tags and word not in my_stopwords] # 길이가 1보다 큰 단어만 반환
```
